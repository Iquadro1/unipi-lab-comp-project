```python
import pickle
import src.bottleneck1 as bc
import matplotlib.pyplot as plt

# Set plot style for better visuals
plt.style.use('seaborn-v0_8-whitegrid')

# Load the pre-computed complex results
with open('results/dragon_complexes.pkl', 'rb') as f:
    complex_res = pickle.load(f)

# The generator notebook ensures the samples are sorted by point count.
sample_names = list(complex_res.keys())
print(f"Analyzing convergence across {len(sample_names)} ordered samples:")
print(sample_names)
# --- Step 1: Compute the convergence sequences ---
convergence_data = bc.compute_convergence_sequences(complex_res)

# --- Step 2: Plot the results on a linear scale ---
# This plot is good for seeing absolute differences.
print("\nPlotting convergence on a linear scale...")
bc.plot_convergence_sequences(convergence_data, sample_names, log_scale=False)

# --- Step 3: Plot the results on a log scale ---
# This plot is better for comparing rates of convergence, especially when values are small.
print("\nPlotting convergence on a log scale...")
bc.plot_convergence_sequences(convergence_data, sample_names, log_scale=True)
```

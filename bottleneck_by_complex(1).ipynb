```python
import pickle
import src.bottleneck1 as bottleneck

# Load the results
with open('results/sphere_complexes.pkl', 'rb') as f:
    complex_res = pickle.load(f)

sample_names = list(complex_res.keys())
complex_types = list(complex_res[sample_names[0]].keys())

# Analyze each complex type across all samples
for complex_type in complex_types:
    # Create a dictionary with sample names as keys and the complex result as values
    complex_comparison = {
        sample_name: complex_res[sample_name][complex_type]
        for sample_name in sample_names
    }

    print(f"\n{'='*80}")
    print(f"ANALYZING BOTTLENECK DISTANCES FOR: {complex_type.upper()} COMPLEX")
    print(f"{'='*80}")

    # Call the single, high-level analysis function
    title = f"Sample Similarity for {complex_type.upper()} Complex"
    bottleneck.analyze_bottleneck_distances(complex_comparison)
```
